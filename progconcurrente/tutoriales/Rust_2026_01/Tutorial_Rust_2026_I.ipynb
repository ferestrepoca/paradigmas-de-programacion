Este código define una *cell magic* personalizada en Google Colab utilizando `IPython`, permitiendo ejecutar código de Rust directamente dentro de una celda mediante el comando `%%rust`.

La función `rust` recibe el contenido completo de la celda y lo escribe directamente en el archivo principal del proyecto de Rust ubicado en `rust_playground/src/main.rs`. De esta forma, cada vez que se ejecuta la celda, el código Rust proporcionado reemplaza el contenido anterior del archivo.

Luego, se utiliza el módulo `subprocess` para ejecutar el comando `cargo run` dentro del directorio del proyecto `rust_playground`. Este comando compila y ejecuta automáticamente el programa Rust definido en `main.rs`.

La salida del proceso se captura a través de `stdout`, incluyendo también los mensajes de error (`stderr` se redirige a stdout), y se imprime línea por línea en tiempo real dentro de la salida de Colab, lo que permite visualizar tanto los resultados del programa como posibles errores de compilación o ejecución de forma inmediata.

Finalmente, el proceso espera a que la ejecución termine mediante `process.wait()`, asegurando que la celda no finalice hasta que el programa Rust haya concluido su ejecución.

````python
from IPython.core.magic import register_cell_magic
import subprocess

@register_cell_magic
def rust(line, cell):

    # Escribir código
    with open(
        "rust_playground/src/main.rs",
        "w"
    ) as f:
        f.write(cell)

    # Ejecutar cargo run
    process = subprocess.Popen(
        ["cargo", "run"],
        cwd="rust_playground",
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    # Mostrar salida en tiempo real
    for linea in process.stdout:
        print(linea, end="")

    process.wait()
````

> **Nota:** Este método puede generar errores de sintaxis al escribir código Rust en Google Colab, ya que las celdas de Colab están diseñadas originalmente para interpretar sintaxis de Python. Debido a esto, el texto del codigo de rust se observa con muchos errores sintacticos porque no es la sintaxis esperada por jupyter.

### Declaración de Variables y Mutabilidad

En RUST las variables se declaran por defecto inmutables utilizando la palabra reservada `let`, a continuación podemos ver un código en el que se hace un uso apropiado de la declaración de variables con `let`.

````python
%%rust

fn main() {
    let x = 10;
    println!("{}", x);
}
````

Sin embargo si usamos solamente `let`, no podremos reasignar otro valor más adelante a la misma variable y obtendremos error.

````python
%%rust

fn main() {
    let x = 10;
    x = 56;
    println!("{}", x);
}
````

````python
%%rust
fn main() {
    let mut contador = 0;
    println!("{}", contador);

    contador += 1;
    println!("{}", contador);


    contador = 56;

    println!("{}", contador);
}
````

Rust deduce automáticamente el tipo de una variable si no se especifica explícitamente.

````python
%%rust

fn main() {
    let edad = 20;       // Rust infiere i32 por defecto
    let nombre = "Ana"; // Rust infiere &str

    println!("Edad: {}", edad);
    println!("Nombre: {}", nombre);
}
````

````python
%%rust

fn main() {
    let mut mi_variable = 10; // Rust infiere i32
    println!("Valor inicial: {}", mi_variable);

    // Esto causará un error de compilación
    // mi_variable = "Hola";
    // println!("Después de asignar string: {}", mi_variable);

    // Esto también causará un error de compilación
    // mi_variable = 20.5;
    // println!("Después de asignar float: {}", mi_variable);

    println!("\nLos comentarios en las líneas de reasignación muestran el error que se produciría.\nRust es un lenguaje de tipado estático, por lo que el tipo de una variable no puede cambiar una vez definido.");
}
````

Como puedes ver en el código comentado, intentar asignar un tipo diferente a una variable ya definida resultaría en un error de compilación. Rust es un lenguaje de tipado estático, lo que significa que el tipo de una variable se fija en tiempo de compilación y no puede cambiar durante la ejecución.

### Declaración de Funciones

#### Sintaxis general

La sintaxis básica para declarar una función en Rust es la siguiente:

```rust
fn nombre_funcion(parametro: Tipo) -> TipoRetorno {
    // código
}
```


#### Ejemplo simple

Una función sencilla que no devuelve ningún valor explícito (devuelve el tipo `()` unit type implícitamente).

````python
%%rust

fn saludar(nombre: &str) {
    println!("Hola {} desde la función!", nombre);
}

fn main() {
    saludar("Carlos");
    saludar("Ana");
}
````

#### Función con retorno

Las funciones pueden devolver un valor. El tipo de retorno se especifica después de la flecha `->`.

````python
%%rust

fn sumar(a: i32, b: i32) -> i32 {
    a + b // La última expresión se retorna implícitamente sin 'return'
}

fn main() {
    let resultado = sumar(5, 3);
    println!("El resultado de la suma es: {}", resultado);

    let otro_resultado = sumar(10, -4);
    println!("Otro resultado: {}", otro_resultado);
}
````

#### Función con múltiples parámetros

Una función puede aceptar tantos parámetros como necesite, cada uno con su propio tipo.

````python
%%rust

fn multiplicar(x: i32, y: i32, z: i32) -> i32 {
    x * y * z
}

fn main() {
    let producto = multiplicar(2, 4, 5);
    println!("El producto es: {}", producto);
}
````

#### Aspectos importantes

*   **Tipado explícito de parámetros:** Todos los parámetros de una función deben declarar su tipo.
*   **Tipado explícito de retorno:** El tipo de retorno de la función debe especificarse explícitamente después de `->`. Si la función no devuelve nada, implícitamente devuelve `()` (el unit type).
*   **Retorno implícito:** La última expresión en el cuerpo de la función puede retornarse sin la palabra clave `return;`. Si terminas una línea con un punto y coma, se convierte en una sentencia y no devolverá un valor, a menos que sea el tipo `()`.

#### Funcion Lambda

Las closures en Rust se utilizan frecuentemente junto con métodos como map() para transformar elementos de una colección de forma compacta y funcional. En el ejemplo map(|x| x * x), la closure recibe cada elemento x y retorna su cuadrado.

````python
%%rust
fn main() {
    let numeros = vec![1, 2, 3, 4];

    let cuadrados: Vec<i32> = numeros
        .iter()
        .map(|x| x * x)
        .collect();

    println!("{:?}", cuadrados);
}
````

#### Relación con concurrencia

Las funciones en Rust manejan la propiedad (`ownership`) o referencias (`borrows`) de los datos que reciben como parámetros. Este sistema es fundamental para la seguridad de la concurrencia en Rust.

Cuando múltiples hilos necesitan acceder a datos compartidos, el compilador de Rust (a través del `borrow checker`) puede verificar que estos accesos sean seguros. Por ejemplo:

*   Si una función recibe la `ownership` de un dato, ningún otro hilo puede poseer o modificar ese dato al mismo tiempo.
*   Si una función recibe una referencia `&` (inmutable) a un dato, múltiples hilos pueden tener referencias inmutables simultáneamente, ya que no modificarán el dato.
*   Si una función recibe una referencia mutable `&mut` a un dato, Rust garantiza que solo puede haber una referencia mutable activa a ese dato en un momento dado, lo que previene condiciones de carrera (`race conditions`) a nivel de acceso a datos mutables.

Este diseño permite que el compilador garantice la seguridad de la memoria y la ausencia de condiciones de carrera relacionadas con el acceso a datos compartidos en tiempo de compilación, sin la necesidad de un recolector de basura o bloqueos de runtime excesivos.

### Tipos Personalizados

#### Structs (Estructuras)

Los `structs` permiten agrupar datos relacionados en una única estructura, lo que es útil para crear tipos complejos que modelan entidades del mundo real.

````python
%%rust

// Definición de un struct Persona
struct Persona {
    nombre: String,
    edad: u32,
}

fn main() {
    // Creación de una instancia de Persona
    let persona = Persona {
        nombre: String::from("Juan"),
        edad: 25,
    };

    // Acceso a los campos del struct
    println!("Nombre: {}", persona.nombre);
    println!("Edad: {}", persona.edad);

    // También se puede crear de forma mutable
    let mut persona_mutable = Persona {
        nombre: String::from("Ana"),
        edad: 30,
    };

    persona_mutable.edad += 1;
    println!("Edad de Ana después de un año: {}", persona_mutable.edad);
}
````

#### Enums (Enumeraciones)

Los `enums` permiten definir un tipo que puede ser uno de varios estados posibles, lo que es ideal para situaciones donde una variable puede tomar un conjunto finito de valores.

````python
%%rust

// Definición de un enum simple
enum EstadoUsuario {
    Activo,
    Inactivo,
    Suspendido,
}

// Enums con datos asociados (tuple structs o anonymous structs)
enum Mensaje {
    Quit, // Sin datos
    Echo(String), // Con un String
    Mover { x: i32, y: i32 }, // Con un struct anónimo
    CambiarColor(i32, i32, i32), // Con tres i32
}

fn procesar_estado(estado: EstadoUsuario) {
    match estado {
        EstadoUsuario::Activo => println!("El usuario está activo."),
        EstadoUsuario::Inactivo => println!("El usuario está inactivo."),
        EstadoUsuario::Suspendido => println!("El usuario ha sido suspendido."),
    }
}

fn main() {
    // Uso de un enum simple
    let usuario1 = EstadoUsuario::Activo;
    let usuario2 = EstadoUsuario::Inactivo;
    procesar_estado(usuario1);
    procesar_estado(usuario2);

    // Uso de enums con datos asociados
    let mensaje1 = Mensaje::Echo(String::from("¡Hola!"));
    let mensaje2 = Mensaje::Mover { x: 10, y: 20 };
    let mensaje3 = Mensaje::CambiarColor(255, 0, 128);

    match mensaje1 {
        Mensaje::Echo(texto) => println!("Mensaje de eco: {}", texto),
        _ => (),
    }

    match mensaje2 {
        Mensaje::Mover { x, y } => println!("Mover a X: {}, Y: {}", x, y),
        _ => (),
    }

    match mensaje3 {
        Mensaje::CambiarColor(r, g, b) => println!("Cambiar color a R:{}, G:{}, B:{}", r, g, b),
        _ => (),
    }
}
````

#### Implementación de Métodos (`impl`)

Podemos asociar funciones (métodos) a nuestros `structs` y `enums` usando bloques `impl`. Estos métodos pueden operar sobre los datos de la instancia.

````python
%%rust

struct Rectangulo {
    ancho: u32,
    alto: u32,
}

// Implementación de métodos para el struct Rectangulo
impl Rectangulo {
    // Un método que calcula el área. Toma una referencia inmutable a self (&self).
    fn area(&self) -> u32 {
        self.ancho * self.alto
    }

    // Un método que crea un nuevo Rectangulo (método asociado, no necesita self)
    fn cuadrado(lado: u32) -> Rectangulo {
        Rectangulo {
            ancho: lado,
            alto: lado,
        }
    }
}

fn main() {
    let r1 = Rectangulo {
        ancho: 10,
        alto: 5,
    };

    // Llamada al método de instancia
    println!("El área del rectángulo es: {}", r1.area());

    let r2 = Rectangulo::cuadrado(7);
    println!("El área del cuadrado es: {}", r2.area());
}
````

#### Relación con Concurrencia (`Send` y `Sync` Traits)

Rust garantiza la seguridad de la concurrencia a través de su sistema de propiedad y préstamo, y rasgos especiales como `Send` y `Sync`.

*   **`Send`**: Un tipo implementa `Send` si se puede mover de un hilo a otro de forma segura. Casi todos los tipos primitivos y la mayoría de los tipos de la biblioteca estándar (incluyendo `String`, `Vec`, `HashMap`) implementan `Send` automáticamente. Si todos los componentes de un `struct` o `enum` implementan `Send`, entonces el tipo compuesto también implementará `Send` automáticamente.

*   **`Sync`**: Un tipo implementa `Sync` si se puede acceder a él a través de una referencia compartida (`&T`) desde múltiples hilos de forma segura. Similar a `Send`, si todos los componentes de un `struct` o `enum` implementan `Sync`, entonces el tipo compuesto también implementará `Sync` automáticamente.

Estos rasgos son marcadores que el compilador utiliza para asegurar que no se introduzcan condiciones de carrera u otros problemas de concurrencia. No necesitas implementarlos manualmente la mayoría de las veces, ya que Rust los deduce automáticamente para tus tipos personalizados si sus componentes son `Send` y `Sync`. Para tipos que gestionan recursos externos o necesitan sincronización explícita, como `Mutex` o `RwLock`, Rust asegura que solo se implementen `Send` y `Sync` de manera que mantengan la seguridad.

## **Manejo de memoria, Ownership y Borrow Checker**

Cada valor en Rust tiene una variable que es su dueño (owner) y solo puede haber un dueño

Cuando la variable dueño sale de scope, el valor correspondiente es eliminado.

Para que otra variable acceda a ese valor, puede copiarlo (de forma profunda con .clone() o de forma superficial implementando el trait Copy), tomarlo prestado mediante una referencia (denominado borrow) o asumir su propiedad (denominado move).

Hacer que en todo momento solo exista un dueño, y que el valor se libere una vez esta variable dueño sale del scope, es lo que permite evitar problemas de gestión de memoria como double free, use after free y memory leaks.

````python
%%rust
//Ejemplo de move
fn main() {
    //La variable s1 es dueña del string "Hola"
    let s1 = String::from("Hola");
    //Se transfiere la propiedad de s1 a s2 y se invalida s1
    let s2 = s1;

    // println!("{}", s1); Error:
    // value borrowed after move

    println!("{}", s2);
}
````

Las variables de tipo entero, float, boolean y tipos simples similares son copiadas en la asignación porque implementan el trait Copy

````python
%%rust
//Ejemplo de copy
fn main() {
    //Rust infiere que x y y son de tipo i32. Este tipo
    //implementa el trait Copy, por lo que la asignación
    //y = x hace que y copie el valor actual de x (5)
    let mut x = 5;
    let y = x;

    //Nota que ambas referencias son válidas, es decir, no se hizo
    //ningún move
    println!("El valor de x es: {}", x);
    println!("El valor de y es: {}", y);

    //Y si modificamos x el valor de y sigue igual,  por lo que se copió
    x = 10;
    println!("-------------------------");
    println!("El valor de x es: {}", x);
    println!("El valor de y es: {}", y);
}
````

Los string literals (es decir, los inicializados de forma estática usando directamente comillas, a diferencia de los inicializados dinámicamente con String::from()) también son copiados en la asignación

````python
%%rust
fn main() {
    let mut s1 = "Hola";
    let s2 = s1;

    println!("{}", s1);
    println!("{}", s2);

    s1 = "Mundo";
    println!("{}", s1);
    println!("{}", s2);
}
````

Podemos hacer copias en lugar de asumir la propiedad de un valor o tomarlo prestado, usando .clone()

````python
%%rust
fn main() {
    let s1 = String::from("Hola");
    let mut s2 = s1.clone();

    //Ambas referencias son válidas, lo cual significa
    //que no se hizo ningún move
    println!("s1 = {}", s1);
    println!("s2 = {}", s2);

    s2.push_str(", Mundo!");
    println!("-------------------");
    println!("s1 = {}", s1);
    println!("s2 = {}", s2);
}
````

La propiedad de los valores se transfiere al llamar funciones

````python
%%rust
fn main() {
    let s1 = String::from("Hola");
    println!("s1 = {}", s1);

    takes_ownership(s1);

    //println!("{}", s1);  Error:
    //value borrowed after move
}

fn takes_ownership(mut some_string: String) {
    some_string.push_str(", Mundo!");
    println!("{}", some_string);
}
````

También se transfiere al retornar de funciones

````python
%%rust
fn main() {
    let mut s1 = String::from("Hola");
    println!("s1 = {}", s1);

    //La propiedad se transfiere de s1 a some_string, y luego
    //se devuelve a s1
    s1 = takes_and_gives_back(s1);

    //La propiedad se transfiere de s1 a some_string. Si ninguna
    //variable la toma de vuelta, el valor es eliminado cuando some_string
    //sale del scope
    //takes_and_gives_back(s1);

    println!("s1 = {}", s1);
}

fn takes_and_gives_back(mut some_string: String) -> String {
    some_string.push_str(", Mundo");
    some_string
}
````

En lugar de tener que transferir la propiedad de un valor de una variable a otra y de función en función, podemos tomarla prestada (borrowing) usando referencias (con &)

````python
%%rust
fn main(){
    let s1 = String::from("Hola");
    let s2 = String::from("Mundo");
    let s1_length = calculate_length_without_borrow(s1);
    let s2_length = calculate_length_with_borrow(&s2);

    println!("s1 = {}", s1);
    println!("Longitud de s1: {}", s1_length);

    println!("s2 = {}", s2);
    println!("Longitud de s2: {}", s2_length);
}

fn calculate_length_without_borrow(some_string: String) -> usize {
    some_string.len()
}

fn calculate_length_with_borrow(some_string: &String) -> usize {
    some_string.len()
}
````

Existen referencias mutables e inmutables (es decir, que pueden modificar al valor subyacente o no). Pueden existir tantas referencias inmutables como deseemos, pero si existe una referencia mutable, no podrá modificar el valor subyacente mientras las referencias inmutables lo estén usando.


Esto es lo que previene condiciones de carrera en Rust

````python
%%rust
fn main() {
    let mut s1 = String::from("Hola");
    //Referencias de solo lectura. Esperan que no se modifique
    //su string subyacente mientras lo usan
    let s2 = &s1;
    let s3 = &s1;

    //No hemos modificado el string, por lo que no hay problema
    println!("s1 = {}", s1);
    println!("s2 = {}", s2);
    println!("Longitud de s2 = {}", calculate_length(s2));
    println!("s3 = {}", s3);

    //Como modificamos el string, ya no podemos usar las referencias
    //s2 y s3
    s1.push_str(", Mundo!");
    println!("-------------------");
    println!("s1 = {}", s1);
    //Estos println ahora lanzarán errores
    //println!("s2 = {}", s2);
    //println!("s3 = {}", s3);
}

fn calculate_length(some_string: &String) -> usize {
    some_string.len()
}
````

### Hilos - Google Collab

Este programa crea **10 tareas concurrentes** utilizando `thread::spawn`, donde cada tarea alterna entre:

1. **Esperar datos (I/O simulado)** usando `thread::sleep`
2. **Realizar cálculos pesados de CPU**

La máquina donde se ejecuta el programa tiene:

- **1 núcleo físico**
- **2 hilos de hardware**

Por ello, aunque se creen 10 threads, el procesador solo puede ejecutar aproximadamente **2 tareas simultáneamente**. Las demás quedan esperando mientras el sistema operativo alterna entre ellas mediante planificación de hilos (*scheduler*).

#### ¿Qué demuestra el ejemplo?

- **Concurrencia:** muchas tareas progresan al mismo tiempo.
- **Paralelismo:** limitado por la cantidad de hilos físicos/lógicos disponibles.
- Mientras una tarea está bloqueada esperando I/O (`sleep`), otra puede usar el CPU.
- El sistema operativo intercambia constantemente los threads activos para aprovechar mejor el procesador.

#### Idea clave

Crear muchos threads **no significa** que todos se ejecuten físicamente al mismo tiempo.  
La concurrencia permite administrar múltiples tareas de forma eficiente, mientras que el paralelismo real depende del hardware disponible.

````python
%%rust
use std::thread;
use std::time::{Duration, Instant};

fn calculo_pesado(limite: u64) {
    let mut acumulador: f64 = 0.0;

    for i in 0..limite {
        acumulador = acumulador * (i as f64);
    }
}

fn ejecutar_tarea(i: u64, n_partes: u64, carga: u64) {
    for j in 0..n_partes {
        // --- Simulando operacion I/O ---
        println!("Tarea {:02} [{:?}] -> Requiriendo datos en disco", i, thread::current().id());
        thread::sleep(Duration::from_millis((100 * i * (n_partes - j)) % 10000));

        // --- PARTE n ---
        println!("Tarea {:02} [{:?}] -> Iniciando PARTE {:02}", i, thread::current().id(), j + 1);
        calculo_pesado(carga);
    }

    println!("Tarea {:02} [{:?}] Trabajo finalizado", i, thread::current().id());
}

fn main() {
    let inicio_programa = Instant::now();
    let mut handles = vec![];

    let n_tareas = 10;
    let n_partes = 5;
    let carga = 50_000_000;

    for i in 0..n_tareas {
        let handle = thread::spawn(move || ejecutar_tarea(i, n_partes, carga));
        handles.push(handle);
    }

    // Esperar a que todos terminen
    for h in handles {
        h.join().unwrap();
    }

    let duracion_total = inicio_programa.elapsed();

    println!("--- Todas las tareas han finalizado. ---");
    println!("DURACIÓN TOTAL DEL PROGRAMA: {:?}", duracion_total);
}
````

###  Lectores y Escritores con Mutex + Arc

Este programa implementa el problema clásico de lectores y escritores usando `Arc` y `Mutex` para compartir un dato entre múltiples hilos.

Se crean varios hilos lectores y un hilo escritor que acceden a un valor compartido. Cada lector lee el valor protegido por el `Mutex`, mientras que el escritor lo modifica sumando 10.

El `Mutex` garantiza que solo un hilo pueda acceder al dato a la vez, evitando condiciones de carrera y corrupción de datos. El `Arc` permite que el mismo recurso sea compartido de forma segura entre todos los hilos.

#### ¿Qué demuestra el ejemplo?

- **Concurrencia:** múltiples hilos ejecutan tareas de lectura y escritura.
- **Exclusión mutua:** el `Mutex` serializa el acceso al dato compartido.
- **Interleaving:** el orden de ejecución depende del scheduler del sistema operativo y no es determinista.

#### Idea clave

Crear múltiples hilos no implica ejecución simultánea real del dato compartido. El `Mutex` asegura seguridad, pero también limita el acceso concurrente al recurso.

````python
%%rust
use std::sync::{Arc, Mutex};
use std::thread;
use std::time::Duration;

fn lector(id: usize, dato: Arc<Mutex<i32>>) {
    for _ in 0..3 {
        let num = dato.lock().unwrap();

        println!("Lector {} lee valor: {}", id, *num);
        thread::sleep(Duration::from_millis(500));
    }
}

fn escritor(dato: Arc<Mutex<i32>>) {
    for _ in 1..=3 {
        {
            let mut num = dato.lock().unwrap();

            *num += 10;
            println!("Escritor modifica valor a: {}", *num);
        }

        thread::sleep(Duration::from_millis(700));
    }
}

fn main() {
    let dato = Arc::new(Mutex::new(0));

    let mut handles = vec![];

    // Lectores
    for i in 0..3 {
        let dato_clonado = Arc::clone(&dato);
        let handle = thread::spawn(move || lector(i, dato_clonado));
        handles.push(handle);
    }

    // Escritor
    let dato_clonado = Arc::clone(&dato);
    let handle_escritor = thread::spawn(move || escritor(dato_clonado));

    handles.push(handle_escritor);

    // Esperar todos los hilos
    for handle in handles {
        handle.join().unwrap();
    }
}
````

### Sensores Concurrentes con Channels

Este programa implementa comunicación concurrente entre múltiples hilos utilizando *channels* del módulo `std::sync::mpsc`.

Se crean dos hilos sensores: uno encargado de generar lecturas de temperatura y otro de humedad. Cada sensor ejecuta su tarea de forma independiente y, cada vez que obtiene una nueva lectura, envía el dato al hilo principal mediante un *channel*.

El hilo principal actúa como una central de monitoreo que recibe y procesa los mensajes enviados por los sensores. Gracias al uso de *channels*, los hilos pueden comunicarse de forma segura sin necesidad de compartir memoria directamente.

#### ¿Qué demuestra el ejemplo?

- **Concurrencia:** múltiples hilos ejecutan tareas de sensado simultáneamente.
- **Paso de mensajes:** los datos se transfieren entre hilos usando `send()` y `recv()`.
- **Comunicación segura:** no existe acceso concurrente a memoria compartida.
- **Interleaving:** el orden de llegada de los mensajes depende del scheduler del sistema operativo y no es determinista.

#### Idea clave

Los *channels* permiten coordinar hilos mediante intercambio de mensajes en lugar de compartir datos directamente, reduciendo la necesidad de mecanismos de sincronización como `Mutex<T>`.

````python
%%rust
use std::sync::mpsc;
use std::thread;
use std::time::Duration;

fn sensor_temperatura(tx: mpsc::Sender<(String, i32)>) {
    let temperaturas = [22, 24, 27, 31, 29];

    for temp in temperaturas {
        println!("[Sensor Temperatura {:?}] Nueva lectura: {}°C", thread::current().id(), temp);

        tx.send(("Temperatura".to_string(), temp)).unwrap();
        thread::sleep(Duration::from_millis(700));
    }
}

fn sensor_humedad(tx: mpsc::Sender<(String, i32)>) {
    let humedades = [40, 45, 50, 48, 43];

    for hum in humedades {
        println!("[Sensor Humedad {:?}] Nueva lectura: {}%", thread::current().id(), hum);

        tx.send(("Humedad".to_string(), hum)).unwrap();
        thread::sleep(Duration::from_millis(500));
    }
}

fn main() {
    let (tx, rx) = mpsc::channel();

    // Sensor temperatura
    let tx_temp = tx.clone();
    let h1 = thread::spawn(move || {
        sensor_temperatura(tx_temp);
    });

    // Sensor humedad
    let tx_hum = tx.clone();
    let h2 = thread::spawn(move || {
        sensor_humedad(tx_hum);
    });

    // Cerrar sender principal
    drop(tx);

    // Centro de monitoreo
    for (tipo, valor) in rx {
        println!("[CENTRAL] Dato recibido -> {}: {}", tipo, valor);
    }

    h1.join().unwrap();
    h2.join().unwrap();

    println!("--- Monitoreo finalizado ---");
}
````

### Thread Pool con Async-Await

Este programa implementa concurrencia masiva utilizando `async/await` y el runtime asíncrono.

Se crean millones de tareas concurrentes que representan clientes conectándose a un servidor. Cada tarea simula espera de red mediante `sleep().await` sin bloquear el hilo de ejecución.

Tokio administra una *thread pool* reutilizando pocos hilos para ejecutar millones de tareas eficientemente.

## ¿Qué demuestra el ejemplo?

- Concurrencia masiva con tareas asíncronas.
- Operaciones no bloqueantes mediante `.await`.
- Uso eficiente de una thread pool.
- Alta escalabilidad comparada con `std::thread`.

## Idea clave

`async/await` permite manejar enormes cantidades de tareas concurrentes sin crear un hilo independiente por cada una, reduciendo consumo de memoria y mejorando la escalabilidad.

````python
%%rust
use tokio::time::{sleep, Duration, Instant};

async fn cliente(id: u32) {
    println!("Cliente {} conectado", id);
    sleep(Duration::from_millis(100)).await;
    println!("Cliente {} desconectado", id);
}

#[tokio::main]
async fn main() {
    let inicio = Instant::now();
    let mut tareas = Vec::new();

    for i in 0..3_000_000 {
        tareas.push(tokio::spawn(cliente(i)));
    }

    for tarea in tareas {
        tarea.await.unwrap();
    }

    let duracion = inicio.elapsed();

    println!("Todos los clientes terminaron");
    println!("Tiempo total: {:?}", duracion);
}
````
