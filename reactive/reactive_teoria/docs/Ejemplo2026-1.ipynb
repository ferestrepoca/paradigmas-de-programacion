# Ejemplo: Monitoreo de chat en tiempo real

Aplicación de chat donde los mensajes de los usuarios llegan constantemente.

El sistema debe mostrar los mensajes normales,
detectar automáticamente mensajes de error,
y generar alertas en tiempo real cuando ocurra un problema.

Este tipo de comportamiento es muy común en:
* aplicaciones de mensajería
* monitoreo de servidores
* notificaciones
* redes sociales

A continuacion se muestra una lista de cadenas de texto de python que tiene el objetivo de simular una lista de mensajes dentro dle contexto dle ejemplo

````python
# Mensajes simulados del chat
messages = [
    "[Juan]: Hola a todos",
    "[Ana]: ¿Cómo están?",
    "[Sistema]: Error de conexión",
    "[Pedro]: Todo bien",
    "[Sistema]: Error al enviar mensaje",
    "[Laura]: Nos vemos luego"
]

````

## implementacion con enfoque no reactivo

Para este primer enfoque el programa controla completamente el flujo de ejecucion y recorre los mensajes almacenados de forma "manual" mediante el bucle for

````python
# Recorrer mensajes manualmente
for msg in messages:

    # Esperar 1 segundo
    time.sleep(1)

    # Verificar si es un error
    if "Error" in msg:
        print(f"⚠ ALERTA -> {msg}")

    else:
        print(f"💬 Nuevo mensaje: {msg}")
````

De la misma forma el programa controla de forma manual los tiempos de ejecucion y de espera mediante el uso del

```
  time.sleep(1)
```
Esto simula lo que se podria interpretar como un sistema que consulta periodicamente y de forma manual

Este enfoque no es reactivo dado que carece de todas las caracteristicas necesarias, tales como flujo de eventos, observables o suscripciones.




# Implementacion con enfoque reactivo

A continuacion se muestra una implementacion reactiva de este contexto en el cual la programacion reactiva funciona especialmente util porque permite trabajar con flujos continuos de eventos y reaccionar automáticamente cuando llegan nuevos datos.

````python
# Emitir un mensaje cada segundo
interval(1.0).pipe(

    # Obtener índice del mensaje
    ops.map(lambda i: messages[i]),

    # Limitar cantidad de mensajes
    ops.take(len(messages))

).subscribe(

    # Reaccionar automáticamente
    lambda msg: (
        print(f"⚠ ALERTA -> {msg}")
        if "Error" in msg
        else print(f"💬 Nuevo mensaje: {msg}")
    )
)

# Mantener el programa vivo
time.sleep(len(messages) + 1)
````

Para empezar, el sistema genera eventos automaticamente con el uso de:


```
interval(1.0)
```
Cada segundo se emite un nuevo evento, simulando la llegada de mensajes en tiempo real.

Esto representa un flujo continuo de datos, es decir **stream**.

De la misma forma el interval al ser la fuente de los eventos, el interval es a su vez el **observable**



**Observer:** Es quien recibe y maneja los eventos emitidos por el observable, en este ejemplo el gragmento de codigo que decide como se manejan los mensajes es:


```
lambda msg: (
        print(f"⚠ ALERTA -> {msg}")
        if "Error" in msg
        else print(f"💬 Nuevo mensaje: {msg}")
    )
```



**Suscriber:** Es quien se suscribe al observable para comenzar a recibir eventos, en este ejemplo el fragmento encargado de suscribirse y escuchar los eventos es:

```
subscribe(...)
```




Ademas se puede observar que para este ejemplo se usa como estrategia de propagacion del cambio **push** en la cual la fuente de los datos envia de forma activa los eventos a sus suscriptores y estos solo escuchan y reaccionan
